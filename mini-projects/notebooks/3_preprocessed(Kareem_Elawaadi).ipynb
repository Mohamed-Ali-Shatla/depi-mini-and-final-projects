```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path



df = pd.read_csv(r"E:\Courses\depi\depi-mini-and-final-projects\mini-projects\data\processed\brazilian_ecommerce_data.csv")

print("Dataset Shape:", df.shape)
display(df.head())
```

```python

df_clean = df.copy()

print("Original shape:", df.shape)
print("Working copy shape:", df_clean.shape)
```

```python

print("Rows and Columns:", df_clean.shape)
print("\nColumn Names:")
print(df_clean.columns.tolist())

print("\nDataset Information:")
df_clean.info()

print("\nStatistical Summary:")
display(df_clean.describe(include="all").T)
```

```python

missing_report = pd.DataFrame({
    "Missing_Count": df_clean.isnull().sum(),
    "Missing_Percentage": (
        df_clean.isnull().mean() * 100
    ).round(2)
}).sort_values("Missing_Percentage", ascending=False)

display(missing_report[missing_report["Missing_Count"] > 0])
```

```python

print("Exact duplicate rows:", df_clean.duplicated().sum())
```

```python

df_clean.columns = (
    df_clean.columns
    .str.strip()
    .str.lower()
    .str.replace(r"\s+", "_", regex=True)
)

print(df_clean.columns.tolist())
```

```python

text_cols = df_clean.select_dtypes(
    include=["object", "string"]
).columns

for col in text_cols:
    df_clean[col] = df_clean[col].map(
        lambda x: x.strip() if isinstance(x, str) else x
    )

df_clean.replace(r"^\s*$", np.nan, regex=True, inplace=True)

print("Text columns cleaned.")
```

```python

before = len(df_clean)

df_clean = df_clean.drop_duplicates().copy()

print("Duplicate rows removed:", before - len(df_clean))
print("Current shape:", df_clean.shape)
```

```python

date_cols = [
    col for col in df_clean.columns
    if ("date" in col or "timestamp" in col)
    and not col.startswith("flag_")
]

for col in date_cols:
    df_clean[col] = pd.to_datetime(
        df_clean[col],
        errors="coerce"
    )

print("Date columns converted:")
print(date_cols)
```

```python

numeric_cols = [
    "price",
    "freight_value",
    "payment_value",
    "payment_installments",
    "payment_sequential",
    "review_score",
    "product_name_length",
    "product_description_length",
    "product_photos_qty",
    "product_weight_g",
    "product_length_cm",
    "product_height_cm",
    "product_width_cm",
    "order_item_id"
]

for col in numeric_cols:
    if col in df_clean.columns:
        df_clean[col] = pd.to_numeric(
            df_clean[col],
            errors="coerce"
        )

print("Numeric conversion completed.")
print(df_clean.dtypes)
```

```python

non_negative_cols = [
    "price",
    "freight_value",
    "payment_value",
    "payment_installments",
    "payment_sequential",
    "product_weight_g",
    "product_length_cm",
    "product_height_cm",
    "product_width_cm",
    "product_name_length",
    "product_description_length",
    "product_photos_qty"
]

for col in non_negative_cols:
    if col in df_clean.columns:
        df_clean.loc[df_clean[col] < 0, col] = np.nan

print("Invalid negative values handled.")
```

```python

if "review_score" in df_clean.columns:
    invalid_reviews = ~df_clean["review_score"].between(1, 5)

    print("Invalid review scores:", invalid_reviews.sum())

    df_clean.loc[invalid_reviews, "review_score"] = np.nan
```

```python

if "product_category" in df_clean.columns:
    df_clean["product_category"] = (
        df_clean["product_category"].fillna("unknown")
    )

print("Missing product categories handled.")
```

```python

outlier_cols = [
    col for col in ["price", "freight_value", "payment_value"]
    if col in df_clean.columns
]

outlier_results = []

for col in outlier_cols:
    values = df_clean[col].dropna()

    if values.empty:
        continue

    Q1 = values.quantile(0.25)
    Q3 = values.quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    count = (
        (df_clean[col] < lower) |
        (df_clean[col] > upper)
    ).sum()

    outlier_results.append({
        "Column": col,
        "Lower_Bound": lower,
        "Upper_Bound": upper,
        "Outlier_Count": int(count)
    })

display(pd.DataFrame(outlier_results))
```

```python

if {
    "order_purchase_timestamp",
    "order_delivered_customer_date"
}.issubset(df_clean.columns):

    df_clean["delivery_days"] = (
        df_clean["order_delivered_customer_date"]
        - df_clean["order_purchase_timestamp"]
    ).dt.total_seconds() / 86400

    df_clean.loc[
        df_clean["delivery_days"] < 0,
        "delivery_days"
    ] = np.nan

    print("Delivery duration feature created.")
else:
    print("Required date columns not found.")
```

```python

if {
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
}.issubset(df_clean.columns):

    df_clean["delivery_delay_days"] = (
        df_clean["order_delivered_customer_date"]
        - df_clean["order_estimated_delivery_date"]
    ).dt.total_seconds() / 86400

    print("Delivery delay feature created.")
else:
    print("Required date columns not found.")
```

```python

if "order_purchase_timestamp" in df_clean.columns:
    df_clean["purchase_year"] = (
        df_clean["order_purchase_timestamp"].dt.year
    )

    df_clean["purchase_month"] = (
        df_clean["order_purchase_timestamp"].dt.month
    )

    df_clean["purchase_day_of_week"] = (
        df_clean["order_purchase_timestamp"].dt.dayofweek
    )

    print("Purchase date features created.")
```

```python

print("Original shape:", df.shape)
print("Cleaned shape:", df_clean.shape)
print("Remaining duplicate rows:", df_clean.duplicated().sum())

print("\nMissing Values:")
display(
    df_clean.isnull().sum()
    .to_frame("Missing_Count")
    .query("Missing_Count > 0")
    .sort_values("Missing_Count", ascending=False)
)

print("\nData Types:")
display(df_clean.dtypes.to_frame("Data_Type"))

print("\nPreview:")
display(df_clean.head())
```

```python

date_cols = [
    "order_purchase_timestamp",
    "order_delivered_customer_date",
    "order_estimated_delivery_date",
    "review_creation_date"
]

for col in date_cols:
    if col in df_clean.columns:
        df_clean[col] = pd.to_datetime(
            df_clean[col],
            errors="coerce"
        )

print(df_clean[date_cols].dtypes)
```

```python

output_file = "E:\Courses\depi\depi-mini-and-final-projects\mini-projects\data\preprocessed\preprocessed.csv"

df_clean.to_csv(
    output_file,
    index=False,
    encoding="utf-8-sig"
)

print("Dataset saved successfully:", output_file)
```
